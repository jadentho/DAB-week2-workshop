```python
import pandas as pd
```

```python
products = pd.read_csv("/Users/jadenthomas/codefold/UE_DAB25/products_wk9_thurs.csv")
sales = pd.read_csv("/Users/jadenthomas/codefold/UE_DAB25/sales_wk9_thurs.csv")
customers = pd.read_csv("/Users/jadenthomas/codefold/UE_DAB25/customers_wk9_thurs.csv")
```

```python
products['launch_date'] = pd.to_datetime(products['launch_date'])
sales['transaction_date'] = pd.to_datetime(sales['transaction_date'])
customers['registration_date'] = pd.to_datetime(customers['registration_date'])
```

```python
sales['year'] = sales['transaction_date'].dt.year
sales['month'] = sales['transaction_date'].dt.month
sales ['day_of_the_week'] = sales['transaction_date'].dt.day_name()
```

```python
sales_merged = sales.merge(
    products[['product_id', 'price', 'category']],
    on='product_id',
    how='left'
)
```
